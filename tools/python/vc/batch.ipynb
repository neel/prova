```python
from varchar import *
```

```python
encoder  = VarCharEncoder(32, 32)
def PaddedCollator(e):

    def collate_batch(batch):
        max_length  = max(len(x) for x in batch)
        indexes     = [torch.tensor(e.encode_chars(x)) for x in batch]
        indexes     = [F.pad(tensor, (0, max_length - tensor.shape[0]), value=0) for tensor in indexes]
        embeddings  = [e(i) for i in indexes]
        expected    = torch.stack(embeddings)
        return expected.to(torch.int32)
    
    return collate_batch

embedder = StrEmbedder()
c = PaddedCollator(embedder)

embeddings = c(["Hello", "World", "Hi"])
embeddings
```
